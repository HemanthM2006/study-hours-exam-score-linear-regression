Study Hours → Exam Score

Dataset:

| **Study Hours (x)** | **Score (y)** |
|:---------------:|:---------:|
| 1               | 35        |
| 2               | 40        |
| 3               | 50        |
| 4               | 55        |
| 5               | 65        |
| 6               | 70        |
| 7               | 80        |
| 8               | 88        |

```python
import numpy as np
import matplotlib.pyplot as plt
```

```python
x = np.array([1., 2., 3., 4., 5., 6., 7., 8.])
y = np.array([35., 40., 50., 55., 65., 70., 80., 88.])
```

```python
# set initial w & b values to 0

w = 0
b = 0
```

```python
def predict(w, x, b):
    return w * x + b
```

```python
def calculate_cost(w, x, b, y):

    m = len(x)

    y_pred = predict(w, x, b)

    cost = (1/ (2 * m)) * np.sum((y_pred - y) ** 2)

    return cost
```

```python
def compute_gradient(w, x, b, y):

    m = len(x)

    y_pred = predict(w, x, b)

    dj_dw = (1/m) * np.sum((y_pred - y) * x)
    dj_db = (1/m) * np.sum(y_pred - y)

    return dj_dw, dj_db
```

```python
iterations = 10000
alpha = 0.01

cost_history = []

for _ in range(iterations):

    dj_dw, dj_db = compute_gradient(w, x, b, y)

    w = w - (alpha * dj_dw)
    b = b - (alpha * dj_db)

    cost = calculate_cost(w, x, b, y)

    cost_history.append(cost)
```

```python
print(calculate_cost(w, x, b, y))
```

```python
print("w:", w)
print("b:", b)
```

```python
plt.plot(cost_history)
plt.xlabel("Iterations")
plt.ylabel("Cost J(w, b)")
plt.title("Cost vs Iterations (Gradient Descent)")
plt.show()
```

```python
y_pred = predict(w, x, b)

plt.scatter(x, y)
plt.plot(x, y_pred)

plt.xlabel("Study Hours")
plt.ylabel("Exam Score")
plt.title("Study Hours vs Exam Score")

plt.show()
```
